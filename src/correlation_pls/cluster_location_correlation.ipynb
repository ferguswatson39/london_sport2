```python
import pandas as pd
import matplotlib.pyplot as plt
import sys
from pathlib import Path
ROOT = Path('/Users/josh/Library/CloudStorage/GoogleDrive-jeverer@gmail.com/My Drive/London Sport/london_sport2')
sys.path.append(str(ROOT / 'src'))
df = pd.read_csv(ROOT / 'exploration' / 'data' / 'master_data' / '2016_to_2023_master_clustering_data_set.csv')
```

```python
from loading_data.data_catalogue import DataCatalogue
dc = DataCatalogue()
borough_labels = dc.get_data_dict()['LA_2023']['value_labels']

cross = pd.crosstab(df['LA_2023'], df['LCA_Class'], normalize='index')
cross.index = [borough_labels.get(i, i) for i in cross.index]

fig, ax = plt.subplots(figsize=(16, 10))
cross.plot(kind='bar', stacked=True, ax=ax, colormap='tab20')
ax.set_xlabel('Borough')
ax.set_ylabel('Proportion')
ax.set_title('LCA Class Distribution by Borough')
ax.legend(title='LCA Class', bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=7)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
