```python
from collections import Counter
import pandas as pd
import numpy as np
import pyreadstat
import re
```

```python
def drop_prefix(df, prefixes):
    cols = [c for c in df.columns if c.startswith(prefixes)]
    df.drop(columns=cols, inplace=True)
```

```python
def load_and_clean(path, year):

    df, meta = pyreadstat.read_sav(path)

    df["year"] = year

    df.rename(columns=rename_map, inplace=True)

    drop_prefix(df, prefixes)

    missing_pct = df.isna().mean()

    high_missing = missing_pct[missing_pct > 0.95]

    cols_to_drop = high_missing.index.difference(["LondInOut"])

    df.drop(columns=cols_to_drop, inplace=True)

    drop_cols = []

    for col in df.columns:
        if re.search(r'_([Q-U])\d+$', col):
            drop_cols.append(col)

    df.drop(columns=drop_cols, inplace=True)

    return df, meta
```

```python
dfs = [df_16_17, df_17_18, df_18_19, df_19_20, df_20_21, df_21_22, df_22_23]

for df in dfs:
    print(df.shape)
```

```python
common_cols = set(dfs[0].columns)

for df in dfs[1:]:
    common_cols &= set(df.columns)

print(len(common_cols))
```

```python
reference = set(df_22_23.columns)

for name, df in zip(["16/17","17/18","18/19","19/20","20/21","21/22","22/23"], dfs):
    
    missing = reference - set(df.columns)

    print(name, len(missing))
```

```python
lost_from_2022 = set(df_22_23.columns) - common_cols

print(len(lost_from_2022))

for col in sorted(lost_from_2022):
    print(col)
```

```python
availability = []

for var in lost_from_2022:

    count = sum(var in df.columns for df in dfs[1:])

    availability.append({
        "variable": var,
        "n_years": count})

availability_df = (
    pd.DataFrame(availability)
    .sort_values("n_years", ascending=False))

print(availability_df)
```

```python
latest_cols = list(df_22_23.columns)

var_counts = Counter()

for df in dfs:
    var_counts.update(df.columns)

keep_cols = [col for col in latest_cols if var_counts[col] >= 1]
```

```python
dfs_aligned = []

for df in dfs:

    df_copy = df[[c for c in df.columns if c in keep_cols]]

    missing_cols = set(keep_cols) - set(df_copy.columns)

    for col in missing_cols:
        df_copy[col] = np.nan

    df_copy = df_copy[keep_cols]

    dfs_aligned.append(df_copy)
```

```python
overall_df = pd.concat(dfs_aligned, axis=0, ignore_index=True)

print(overall_df.shape)
```

```python
overall_df = overall_df[overall_df['LondInOut'].notna()]

overall_df["calendar_month"] = ((overall_df["month"] - 3) % 12) + 1
overall_df["calendar_year"] = overall_df["year"].str.split("/").str[1].astype(int) + 2000
overall_df.loc[overall_df["calendar_month"].isin([11, 12]), "calendar_year"] = overall_df["calendar_year"] - 1

overall_df['MEMS7_ALL'] = overall_df['MEMS7_ALL'].clip(upper = 6720)

overall_df['active'] = overall_df['MEMS7_ALL'] >= 150

# overall_df = pyreadstat.set_value_labels(overall_df, meta_22_23)
```
