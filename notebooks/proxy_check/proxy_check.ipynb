```python
plt.scatter(df['IMD10'], df['status'])
```

```python
x = df['IMD10'].values
y = df['status'].values
coefs = np.polyfit(x, y, deg=1)
poly = np.poly1d(coefs)
plt.scatter(x, y, c=df['simpsons_index'].values)
plt.plot(x, poly(x))
```
